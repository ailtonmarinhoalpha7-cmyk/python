```python
produtos = ["Teclado", "Mouse", "Monitor"]
print("=== ESTOQUE FÁCIL ===")
print("1 - Listar produtos")
print("2 - Cadastrar produtos")
print("3 - Atualizar produto")
print("4 - Deletar produto")
opcao = input("Escolha uma opção:")
while opcao !="0":
  #---------------------------------------------------------------
  # LISTAR PRODUTOS
  if opcao == "1":
    print("Produtos Cadastrados", produtos)
  #-------------------------------------------------------------
  # CADASTRAR PRODUTO
  elif opcao == "2":
      novo_produto = input("Digite o nome do produto: ")
      produtos.append(novo_produto)
      print("Produto cadastrado com sucesso!")
      print(produtos)

  #---------------------------------------------------------------
  # ATUALIZAR PRODUTO
  # Atualizar produto pelo nome usando index()

  elif opcao == "3":
      produto_antigo = input("Digite o nome do produto que deseja atualizar: ")
      if produto_antigo in produtos:
          index = produtos.index(produto_antigo)
          novo_produto = input("Digite o novo nome do produto: ")
          produtos[index] = novo_produto
          print("Produto atualizado com sucesso!")
          print(produtos)
      else:
          print("Produto não encontrado!")

  #---------------------------------------------------------------
  # DELETAR PRODUTO
  elif opcao == "4":
      produto_excluir = input("Digite o nome do produto que deseja excluir: ")
      if produto_excluir in produtos:
          produtos.remove(produto_excluir)
          print("Produto excluído com sucesso!")
          print(produtos)
      else:
          print("Produto não encontrado")

  else:
      print("Opção inválida!")

  opcao = input("Escolha uma opção")
```
